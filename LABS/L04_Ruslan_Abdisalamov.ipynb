````python
#@title Run me first &mdash; Lab 4 checkpoint system  { display-mode: "form" }
# You do not need to read or change this cell. Just run it once, before anything else.
# Type your name in the box on the right (in Colab) or edit STUDENT_NAME below.

STUDENT_NAME = "Ruslan Abdisalamov"  #@param {type:"string"}

import io, contextlib
from IPython import get_ipython

LAB = "Lab 4"
_RESULTS = {}
_ORDER = ['Q1', 'Q2', 'Q3', 'Q4', 'Q5', 'Q6']
_SPEC = {'Q1': {'title': 'Comparisons that produce True or False',
        'hint': "Each answer must be a comparison, not a fixed True/False. 'At least 10' means >= "
                '10, not > 10.',
        'cases': [{'inputs': {'x': 7, 'y': 10},
                   'vars': {'is_x_smaller': True,
                            'are_equal': False,
                            'is_y_even': True,
                            'is_x_at_least_10': False}},
                  {'inputs': {'x': 10, 'y': 10},
                   'boundary': True,
                   'vars': {'is_x_smaller': False,
                            'are_equal': True,
                            'is_y_even': True,
                            'is_x_at_least_10': True}},
                  {'inputs': {'x': 12, 'y': 7},
                   'vars': {'is_x_smaller': False,
                            'are_equal': False,
                            'is_y_even': False,
                            'is_x_at_least_10': True}},
                  {'inputs': {'x': 0, 'y': 0},
                   'vars': {'is_x_smaller': False,
                            'are_equal': True,
                            'is_y_even': True,
                            'is_x_at_least_10': False}}]},
 'Q2': {'title': 'A one-way if statement',
        'hint': "The warning is printed only sometimes, but 'Have a safe trip.' is printed every "
                'time - so it must NOT be indented.',
        'cases': [{'inputs': {'speed': 70}, 'out': ['Slow down!', 'Have a safe trip.']},
                  {'inputs': {'speed': 65}, 'boundary': True, 'out': ['Have a safe trip.']},
                  {'inputs': {'speed': 30}, 'out': ['Have a safe trip.']},
                  {'inputs': {'speed': 66},
                   'boundary': True,
                   'out': ['Slow down!', 'Have a safe trip.']}]},
 'Q3': {'title': 'An if-else statement: even or odd',
        'hint': 'A number is even when the remainder after dividing by 2 is zero:  number % 2 == 0',
        'cases': [{'inputs': {'number': 8}, 'out': ['Even']},
                  {'inputs': {'number': 7}, 'out': ['Odd']},
                  {'inputs': {'number': 0}, 'boundary': True, 'out': ['Even']},
                  {'inputs': {'number': -3}, 'out': ['Odd']}]},
 'Q4': {'title': 'Storing the result of a decision',
        'hint': 'A score of exactly 60 passes, so the comparison needs >= and not >.',
        'cases': [{'inputs': {'score': 60},
                   'boundary': True,
                   'vars': {'result': 'Pass'},
                   'out': ['Result: Pass']},
                  {'inputs': {'score': 59},
                   'boundary': True,
                   'vars': {'result': 'Fail'},
                   'out': ['Result: Fail']},
                  {'inputs': {'score': 100}, 'vars': {'result': 'Pass'}, 'out': ['Result: Pass']},
                  {'inputs': {'score': 0}, 'vars': {'result': 'Fail'}, 'out': ['Result: Fail']}]},
 'Q5': {'title': 'Find and fix three bugs',
        'hint': 'There are exactly three problems: a missing colon, a missing colon on else, and a '
                'line that is not indented.',
        'cases': [{'inputs': {'age': 12}, 'out': ['Ticket: $8', 'Enjoy the movie!']},
                  {'inputs': {'age': 13},
                   'boundary': True,
                   'out': ['Ticket: $12', 'Enjoy the movie!']},
                  {'inputs': {'age': 5}, 'out': ['Ticket: $8', 'Enjoy the movie!']},
                  {'inputs': {'age': 40}, 'out': ['Ticket: $12', 'Enjoy the movie!']}]},
 'Q6': {'title': 'Challenge: free shipping over $50',
        'hint': 'Two separate decisions: an if-else that sets shipping, then a one-way if that '
                'prints the free-shipping message.',
        'cases': [{'inputs': {'total': 62},
                   'vars': {'shipping': 0, 'grand_total': 62},
                   'out': ['Free shipping applied!', 'Shipping: 0', 'Grand total: 62']},
                  {'inputs': {'total': 50},
                   'boundary': True,
                   'vars': {'shipping': 0, 'grand_total': 50},
                   'out': ['Free shipping applied!', 'Shipping: 0', 'Grand total: 50']},
                  {'inputs': {'total': 49},
                   'boundary': True,
                   'vars': {'shipping': 6, 'grand_total': 55},
                   'out': ['Shipping: 6', 'Grand total: 55']},
                  {'inputs': {'total': 0},
                   'vars': {'shipping': 6, 'grand_total': 6},
                   'out': ['Shipping: 6', 'Grand total: 6']}]}}


def _find_source(qid):
    """Find the most recent cell the student ran that carries this question's tag."""
    ip = get_ipython()
    hist = list(ip.user_ns.get("In", [])) if ip else []
    tag = "# --- " + qid + " ---"
    for src in reversed(hist[:-1]):
        if tag in src:
            return src
    return None


def _blocked_input(*args, **kwargs):
    raise RuntimeError(
        "Do not use input() in this lab. The values you need are given to you in the cell above."
    )


def _run(src, inputs):
    ns = dict(inputs)
    ns["input"] = _blocked_input
    ns["__name__"] = "__main__"
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        exec(compile(src, "<your answer>", "exec"), ns)
    return ns, buf.getvalue()


def _same(got, want):
    if isinstance(want, bool):
        return got is want
    if isinstance(want, (int, float)) and isinstance(got, (int, float)) and not isinstance(got, bool):
        return abs(got - want) < 1e-9
    return got == want


def _show(value):
    return "<never set>" if value is _MISSING else repr(value)


class _Missing:
    def __repr__(self):
        return "<never set>"


_MISSING = _Missing()


def _evaluate(qid):
    """Returns (status, report_lines)."""
    spec = _SPEC.get(qid)
    if spec is None:
        return "not attempted", ["Unknown question id: " + qid]

    src = _find_source(qid)
    if src is None:
        return "not attempted", [
            "I could not find your answer.",
            "Run the cell that begins with   # --- " + qid + " ---   then run this checkpoint again.",
            "(Keep that comment line at the top of your answer cell.)",
        ]

    lines, passed = [], 0
    for case in spec["cases"]:
        label = ", ".join(str(k) + " = " + repr(v) for k, v in case["inputs"].items())
        if case.get("boundary"):
            label += "   (boundary value)"
        try:
            ns, out = _run(src, case["inputs"])
        except Exception as exc:
            lines.append("  [X] " + label)
            lines.append("        your code stopped with " + type(exc).__name__ + ": " + str(exc))
            continue

        problems = []
        for name, want in case.get("vars", {}).items():
            got = ns.get(name, _MISSING)
            if got is _MISSING or not _same(got, want):
                problems.append(
                    "variable " + name + ": expected " + repr(want) + ", your code gave " + _show(got)
                )
        if "out" in case:
            want_lines = case["out"]
            got_lines = [l.rstrip() for l in out.strip().splitlines() if l.strip()]
            if got_lines != want_lines:
                problems.append("the printed output does not match")
                problems.append("    expected: " + (" | ".join(want_lines) if want_lines else "(nothing printed)"))
                problems.append("    you got:  " + (" | ".join(got_lines) if got_lines else "(nothing printed)"))

        if problems:
            lines.append("  [X] " + label)
            for p in problems:
                lines.append("        " + p)
        else:
            passed += 1
            lines.append("  [OK] " + label)

    total = len(spec["cases"])
    if passed == total:
        lines.append("")
        lines.append("CORRECT - all " + str(total) + " test cases passed.")
        return "passed", lines
    lines.append("")
    lines.append("NOT YET - " + str(passed) + " of " + str(total) + " test cases passed.")
    lines.append("Look at the [X] rows above: they tell you which input value your code gets wrong.")
    if spec.get("hint"):
        lines.append("Hint: " + spec["hint"])
    return "failed", lines


def check(qid):
    """Checkpoint for one question. Run this after your answer cell."""
    status, lines = _evaluate(qid)
    _RESULTS[qid] = status
    title = _SPEC[qid]["title"] if qid in _SPEC else ""
    bar = "-" * 68
    print(bar)
    print("CHECKPOINT " + qid + "  |  " + title)
    print(bar)
    for l in lines:
        print(l)
    print(bar)


def final_check():
    """Re-checks every question from scratch and prints one summary."""
    marks = {"passed": "PASSED", "failed": "NOT CORRECT YET", "not attempted": "NOT ATTEMPTED"}
    codes, done = [], 0
    rows = []
    for qid in _ORDER:
        buf = io.StringIO()
        with contextlib.redirect_stdout(buf):
            status, _ = _evaluate(qid)
        _RESULTS[qid] = status
        rows.append((qid, _SPEC[qid]["title"], marks[status]))
        codes.append(qid + ":" + {"passed": "P", "failed": "F", "not attempted": "-"}[status])
        done += status == "passed"

    n = len(_ORDER)
    bar = "=" * 68
    print(bar)
    print("CS 250  |  " + LAB + "  |  FINAL CHECK")
    print("Student: " + (STUDENT_NAME if STUDENT_NAME else "(no name entered)"))
    print(bar)
    for qid, title, mark in rows:
        print("  " + qid + "  " + (title + " ").ljust(44, ".") + " " + mark)
    print("-" * 68)
    if done == n:
        print("ALL " + str(n) + " QUESTIONS CORRECT. Great work - you are finished.")
    else:
        print(str(done) + " of " + str(n) + " questions correct.")
        print("Go back to any question not marked PASSED, fix it, and run final_check() again.")
    print()
    print("COPY THIS LINE INTO CANVAS:")
    print("  " + LAB.upper().replace(" ", "") + " | " + (STUDENT_NAME if STUDENT_NAME else "NO NAME")
          + " | " + " ".join(codes) + " | " + str(done) + "/" + str(n))
    print(bar)


print(LAB + " checkpoint system loaded. " + str(len(_ORDER)) + " questions to complete.")
print("After each answer cell, run its check(\"Qn\") cell. At the end, run final_check().")
````

## Guided example 1 &nbsp;·&nbsp; Questions that Python can answer

A **Boolean value** is one of exactly two pieces of data: `True` or `False`.
A **Boolean expression** is a question Python works out, and the answer is always one of those two values.

Run the cell, then change the numbers and run it again.

````python
age = 21
is_enrolled = True

print(age >= 18)          # a comparison -> True
print(age == 21)          # equal to?    -> False
print(age != 21)          # not equal?   -> True
print(age % 2 == 0)       # is age even? -> True

print(is_enrolled)        # already a Boolean, no comparison needed
print(type(is_enrolled))  # <class 'bool'>

# = stores a value, == asks a question. They are not the same thing.
score = 90                # store 90 in score
print(score == 90)        # ask whether score is 90  -> True
print(score)              # score is unchanged       -> 90
````

## Guided example 2 &nbsp;·&nbsp; The one-way `if`

A one-way `if` does something **only sometimes**. When the condition is `False`, Python simply skips the block
and carries on &mdash; nothing breaks.

Four parts are required, and leaving out any one of them stops the program running:

1. the keyword `if`
2. a Boolean expression (the condition)
3. a colon `:` at the end of the line
4. an **indented** block underneath &mdash; four spaces

Run this, then change `attempts` to `2` and run it again. Watch which lines disappear.

````python
attempts = 4

if attempts > 3:
    print("Account locked.")
    print("Contact the help desk.")

print("Session ended.")

# The two indented lines belong to the if.
# "Session ended." starts at the left margin, so it always runs.
````

## Guided example 3 &nbsp;·&nbsp; The two-way `if-else`

An `if-else` has **exactly two** outcomes. One block runs, the other does not &mdash; never both, never neither.

`else` takes no condition of its own, needs its own colon, and must line up underneath its `if`.

````python
temperature = 92

if temperature > 85:
    print("Stay hydrated!")
else:
    print("Pleasant weather.")

print("Have a great day.")

# Trace it: 92 > 85 is True -> the first block runs, the else block is skipped.
# Change temperature to 60 and run again: now only the else block runs.
# Change it to exactly 85: 85 > 85 is False, so the else block runs. Boundary values matter.
````
